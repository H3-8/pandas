```python
#Exo 1.1 -  Crée une colonne mois et donne le CA total par mois. 👀
df['month'] = df['date_commande'].dt.month
df.groupby('month')['total_achat'].sum()
```

```python
#Exo 1.2 - Pour chaque ville, donne le nombre de clients distincts et leur dépense moyenne. 👀

summary_df = df.groupby('ville').agg(
    unique_client=('client_id', 'nunique'),
    mean_spending=('total_achat', 'mean')
).round(2)
```

```python
summary_df['mean_spending'] = summary_df['mean_spending'].astype(str) + ' €'
summary_df
```

```python
#Exo 1.3 - Identifie les villes avec une dépense moyenne au-dessus de la médiane globale de total_achat. 👀

med_city_above_average = df.groupby('ville')['total_achat'].mean()
med_city_above_average = med_city_above_average[med_city_above_average > df['total_achat'].median()]
med_city_above_average
```

```python
#Exo 2.1 - Donne le mois avec le plus fort taux de clients VIP (en % du total des commandes de ce mois). 👀

vip = df[df['statut'] == 'VIP'].groupby('month')['quantite'].count()
vip_percent_by_month = (vip / df.groupby('month')['quantite'].count() * 100).round(2)
vip_percent_by_month.nlargest(1)
```

```python
#Exo 2.2 - Donne la répartition des notes de satisfaction par statut en %. 👀


satisfaction = df.groupby(['statut', 'note_satisfaction']).size()
satisfaction
rating_by_status = satisfaction / satisfaction.groupby('statut').transform('sum') * 100
rating_by_status.round(2)
```

```python
#Exo 3.2 - Crée une colonne decennie qui regroupe les ages par tranche : 18-29, 30-39, 40-49, 50+. Ensuite donne la dépense moyenne par tranche. 👀


df['decennie'] = pd.cut(
    df['age'],
    bins=[18, 29, 39, 49, float('inf')],
    labels=['18-29', '30-39', '40-49', '50+']
)
df.groupby('decennie')['total_achat'].mean()
```

```python
#Exo 3.3 - Parmi les clients de 50+, quels pays ont une dépense médiane supérieure à la médiane globale de ce groupe ? 👀

filtre = df[df['decennie'] == '50+'].groupby('pays')['total_achat'].median()
pays_seniors_high_spend = filtre[filtre > df[df['decennie'] == '50+']['total_achat'].median()
]
pays_seniors_high_spend
```

```python
#Exo 3.4 - Applique un rank() sur total_achat par pays — du plus dépensier au moins — et affiche les clients qui sont #1 dans leur pays. 👀

df['rank_in_pays'] = df.groupby('pays')['total_achat'].rank(ascending=False, method='dense')
df[df['rank_in_pays'] == 1][['client_id', 'pays', 'total_achat', 'rank_in_pays']]
```
